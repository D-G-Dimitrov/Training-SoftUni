```python
from google.colab import userdata

# NotionDBLoader: loads pages from a Notion database as LangChain Documents.
# This is useful for building RAG (Retrieval-Augmented Generation) pipelines
# that can answer questions using your Notion content.
from langchain_community.document_loaders import NotionDBLoader

# Retrieve credentials from Colab secrets:
# - NOTION_TOKEN: the Notion integration token (created in Notion's developer settings)
# - NOTION_DATABASE_ID: the ID of the specific Notion database to load from
#   (found in the database's URL: notion.so/your-workspace/<DATABASE_ID>?v=...)
notion_token = userdata.get('NOTION_TOKEN')
notion_database_id = userdata.get('NOTION_DATABASE_ID')
```

```python
# Create the Notion loader by providing the integration token and database ID.
# The loader will use the Notion API to fetch all pages in the specified database.
notion_loader = NotionDBLoader(integration_token=notion_token, database_id=notion_database_id)
```

```python
# Fetch all pages from the Notion database.
# Each page is returned as a LangChain Document with:
#   - page_content: the text content of the page
#   - metadata: additional info like page title, page ID, last edited time, etc.
notion_loader.load()
```
