```python
# Import Google Colab utilities for reading secrets
from google.colab import userdata

# GithubFileLoader: loads file contents from a GitHub repository as LangChain Documents
# GitHubIssuesLoader: loads GitHub issues as LangChain Documents (useful for RAG pipelines)
from langchain_community.document_loaders import GithubFileLoader, GitHubIssuesLoader

# Retrieve the GitHub personal access token from Colab secrets.
# This token is used to authenticate with the GitHub API (needed for private repos or higher rate limits).
github_token = userdata.get('GITHUB_TOKEN')
```

```python
# Configure a loader that will fetch files from a GitHub repository.
# - repo: the "owner/repository" path on GitHub
# - access_token: GitHub personal access token for authentication
# - file_filter: a lambda (anonymous function) that filters which files to load.
#   Here we keep only .md (Markdown) files and skip paths containing '#'
#   (GitHub encodes some special paths with '#' which can cause issues).
github_file_loader = GithubFileLoader(
    repo="TonyTroeff/Training-SoftUni",
    access_token=github_token,
    file_filter=lambda path: path.find('#') == -1 and path.endswith(".md")
)
```

```python
# Load all matching files from the repository.
# Each file is returned as a LangChain Document object with:
#   - page_content: the file's text content
#   - metadata: info like file path, repository, etc.
github_file_loader.load()
```

```python
# Configure a loader that fetches GitHub Issues from a repository.
# Issues are returned as LangChain Documents — great for building Q&A systems over issue trackers.
# - repo: the target GitHub repository
# - access_token: for authenticated requests (higher API rate limits)
# - per_page: how many issues to fetch per API page (max 100)
github_issues_loader = GitHubIssuesLoader(
    repo="TryAtSoftware/CleanTests",
    access_token=github_token,
    per_page=50
)
```

```python
# Load all issues from the repository.
# Returns a list of Document objects, one per issue, with title and body as content.
github_issues_loader.load()
```
